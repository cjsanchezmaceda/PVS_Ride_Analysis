# Part 1 - PVS Dataset Exploration

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
file = Path("../data/raw/PSV1")
data = {}
for csv in file.glob("*.csv"):
    data[csv.stem] = pd.read_csv(csv, nrows=1000)
    print(csv.name, data[csv.stem].shape)
    print(list(data[csv.stem].columns))
    display(data[csv.stem].head())
```

```python
for name in data:
    if "timestamp" in data[name].columns:
        dt=data[name]["timestamp"].diff()
        fs = 1/dt
        fs_nyquist = 0.5 * fs.max()
        print(name, round(fs.median(),2), "Hz", round(fs_nyquist,2), "Hz")
```

```python
path_gps_mpu_left="../data/raw/PSV1/dataset_gps_mpu_left.csv"
path_dataset_labels="../data/raw/PSV1/dataset_labels.csv"
gps_mpu_left=pd.read_csv(path_gps_mpu_left)
dataset_labels=pd.read_csv(path_dataset_labels)
print(len(gps_mpu_left), len(dataset_labels))
print(len(gps_mpu_left)==len(dataset_labels))
```

```python
path_gps_mpu_right="../data/raw/PSV1/dataset_gps_mpu_right.csv"
gps_mpu_right=pd.read_csv(path_gps_mpu_right)

left_data=pd.concat([gps_mpu_left, dataset_labels], axis=1)
right_data=pd.concat([gps_mpu_right, dataset_labels], axis=1)

print(list(left_data.columns))
display(left_data.head())
print(list(right_data.columns))
display(right_data.head())
```

```python
left_data['t']=left_data['timestamp']-left_data['timestamp'].iloc[0]
right_data['t']=right_data['timestamp']-right_data['timestamp'].iloc[0]
print(left_data['t'].head())
print(right_data['t'].head())

left_data['dt']=left_data['t'].diff()
right_data['dt']=right_data['t'].diff()

print(left_data['dt'].max())
print(right_data['dt'].max())
```

```python
t0_left=left_data.loc[left_data['speed']>10, 't'].iloc[0]
interval_left=left_data[(left_data['t']>=t0_left) & (left_data['t']<t0_left+30)]

t0_right=right_data.loc[right_data['speed']>10, 't'].iloc[0]
interval_right=right_data[(right_data['t']>=t0_right) & (right_data['t']<t0_right+30)]

display(interval_left)
display(interval_right)
```

```python
road=['asphalt_road', 'cobblestone_road', 'dirt_road']
right_data['surface']=right_data[road].idxmax(axis=1)

fig, axes=plt.subplots(2,1, figsize=(7,5), sharex=True)
axes[0].plot(right_data['t'], right_data['speed'])
axes[0].set_ylabel('Speed (m/s)')

axes[1].plot(left_data['t'], right_data['surface'])
axes[1].set_ylabel('Surface Type')
axes[1].set_xlabel('Time (s)')

plt.tight_layout()
plt.show()
```

```python
road=['asphalt_road', 'cobblestone_road', 'dirt_road']
interval_left['surface']=interval_left[road].idxmax(axis=1)

fig, axes=plt.subplots(4,1, figsize=(10,8), sharex=True)
axes[0].plot(interval_left['t'], interval_left['acc_z_below_suspension'], label='below_suspension')
axes[0].plot(interval_left['t'], interval_left['acc_z_above_suspension'], label='above_suspension')
axes[0].legend()
axes[0].set_ylabel('Acceleration (m/s^2)')

axes[1].plot(interval_left['t'], interval_left['acc_z_dashboard'])
axes[1].set_ylabel('Acceleration (m/s^2)')
axes[1].set_ylim(axes[0].get_ylim())

axes[2].plot(interval_left['t'], interval_left['speed'])    
axes[2].set_ylabel('Speed (m/s)')

axes[3].plot(interval_left['t'], interval_left['surface'])
axes[3].set_ylabel('Surface Type')
axes[3].set_xlabel('Time (s)')

plt.tight_layout()
plt.show()
```

```python
fig, axes=plt.subplots(4,1, figsize=(10,8), sharex=True)
interval_right_roughsurf=right_data[(right_data['t']>=300) & (right_data['t']<450)]

axes[0].plot(interval_right_roughsurf['t'], interval_right_roughsurf['acc_z_below_suspension'], label='below_suspension')
axes[0].plot(interval_right_roughsurf['t'], interval_right_roughsurf['acc_z_above_suspension'], label='above_suspension')
axes[0].legend()
axes[0].set_ylabel('Acceleration (m/s^2)')

axes[1].plot(interval_right_roughsurf['t'], interval_right_roughsurf['acc_z_dashboard'])
axes[1].set_ylim(axes[0].get_ylim())
axes[1].set_ylabel('Acceleration (m/s^2)')

axes[2].plot(interval_right_roughsurf['t'], interval_right_roughsurf['speed'])    
axes[2].set_ylabel('Speed (m/s)')

axes[3].plot(interval_right_roughsurf['t'], interval_right_roughsurf['surface'])
axes[3].set_ylabel('Surface Type')
axes[3].set_xlabel('Time (s)')

plt.tight_layout()
plt.show()
```

# Conslusions
1. Sensors are mounted on the front suspension: above (body, top of the strut tower – sprung mass) and below (lower control arm – unsprung mass), plus the dashboard 
2. IMU data is sampled at 100 Hz with no timestamp gaps (Nyquist frequency 50 Hz), sufficient for the expected sprung-mass (~1–2 Hz) and unsprung-mass (~10–15 Hz) modes. GPS data is logged at 1 Hz, so speed appears as 1 s steps.
3. The z-axis is vertical in all three positions; the mean vertical acceleration is ~ 9.81 (gravity) everywhere.
4. The standard deviation of vertical acceleration below the suspension is 3.3–3.7 times higher than above it, depending on the surface. The dashboard shows the lowest vibration of the three positions.
5. Dirt produces the highest vibration below the suspension (std 9.46 m/s^2 vs 8.55 m/s^2 on cobblestone), while cobblestone transmits the most vibration to the dashboard (2.34 vs 2.05 m/s^2). Hypothesis to test: the two surfaces excite different frequency ranges, which the suspension attenuates differently.
6. Dirt and cobblestone sections were driven at lower speeds than asphalt.
7. Short frozen segments were found (e.g. t ~ 132–133 s, constant values in all sensors) and peaks approach the +-8 g sensor range. Both need to be detected and excluded in Part 2.
8. These statistics include stationary and low-speed periods and frozen samples, so they are indicative only; Part 2 will compute them on clean, constant-speed windows.
